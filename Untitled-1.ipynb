```python
import csv
import yaml
import math
import numpy as np
from pathlib import Path
#引用内容
```

```python
#定义读取yaml的函数
def read_yaml(comein):
  text=Path(comein)
  if not text.exists():
    print("错误，yaml文件不存在,程序停止")
    exit()
  f=open(text,"r",encoding="utf-8")
  inyaml=yaml.safe_load(f)
  f.close()
  if inyaml is None:
      print("任务内容为空")
  print("任务要求：")
  print(inyaml)
```

```python
def read_csv(csv):
    word = Path(csv)
    if not word.exists():
        print("错误,csv文件不存在,程序停止")
        exit()
    f = open(word, "r", encoding="utf-8", newline="")
    reader = csv.reader(f)
    header = next(reader, None)
    rows = [row for row in reader]
    f.close()
    if header is None:
        print("错误，csv没有表头，程序停止")
        exit()
    if len(header) < 2:
        print("错误，列名不匹配：期望两列")
        exit()
    if not rows:
        print("错误，csv为空或没有数据行")
        exit()
    x_data = [float(row[0]) for row in rows]
    y_data = [float(row[1]) for row in rows]
    return x_data, y_data
```

```python
#相关系数方程
def r(x, y):
    n = len(x)
    if n == 0:
        print("错误，数据为空，程序停止")
        exit()
    if len(x) != len(y):
        print("错误，两列长度不一致，程序停止")
        exit()

    average_x = sum(x) / n
    average_y = sum(y) / n

    s_xy = sum((xi - average_x) * (yi - average_y) for xi, yi in zip(x, y))
    s_xx = sum((xi - average_x) ** 2 for xi in x)
    s_yy = sum((yi - average_y) ** 2 for yi in y)

    dividend= math.sqrt(s_xx * s_yy)
    if dividend == 0:
        print("错误,某列方差为0,程序停止")
        exit()
    return s_xy/ dividend
```

```python
#验证结果
def vs(x,y):
     my_r = r(x,y)
     np_r = np.corrcoef(x, y)[0,1]
     if abs(my_r - np_r) < 1e-10:
         print(f"结果正确，r = {my_r:.2_____f}")
     else:
         print("错误")
```

```python
#主体内容
def main():
    task=read_yaml("yaml的地址")
    x_data, y_data = read_csv("csvd的地址")
    me_r = r(x_data, y_data)
    print(f"计算得到相关系数 r = {me_r:.2f}")
    vs(x_data, y_data)
```

```python
if __name__ =="__main__":
    main()
#防止引用时直接启动程序
```
